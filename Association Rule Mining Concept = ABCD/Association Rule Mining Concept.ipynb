```python
data= {
    "C1":["A",'B','C','D','E'],
    "C2":["A",'B','F','G','H'],
    "C3":["N",'Y','C','Z','X'],
    "C4":["F",'C','S','D','E'],
    "C5":["Q",'B','K','V','G'],
    "C6":["A",'F','R','W','E'],
    "C7":["A",'L','C','J','E']
    
}
```

```python
import pandas as pd
```

```python
D=pd.DataFrame(data)
D
```

```python
TR=D.values.tolist()
TR
```

```python
# step 1: Convert columns to Transactions (rows)
transaction=list(zip(*data.values()))
```

```python
transaction
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
te=TransactionEncoder()
```

```python
transaction
```

```python
# step 2: Use TransectionEncoder

te_array=te.fit(transaction).transform(transaction)
te_array
```

```python
df=pd.DataFrame(te_array,columns=te.columns_)
df
```

```python
# setp 3: Find frequent itemsets
frequent_items=apriori(df,min_support=0.3,use_colnames=True,verbose=1)
```

```python
frequent_items
```

```python
# step 4: Generate association Rule

rule=association_rules(frequent_items,metric='confidence',min_threshold=0.6)
```
