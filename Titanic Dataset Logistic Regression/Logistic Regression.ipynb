```python
data=df[['Pclass',"Sex","Age","SibSp","Parch","Fare","Embarked","Survived"]]
```

```python
data["Family_Size"]=data['SibSp']+data["Parch"] +1
```

```python
data.info()
```

```python
data['Age']=data['Age'].fillna(round(data["Age"].mean(),0)).astype(int)
```

```python
A=data["Embarked"].mode()[0]
```

```python
data["Embarked"].fillna(A,inplace=True)
```

```python
data.info()
```

```python
data.duplicated().sum()
```

```python
data.reset_index(drop=True)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder=LabelEncoder()
```

```python
data["Sex"]=encoder.fit_transform(data["Sex"])
data["Embarked"]=encoder.fit_transform(data["Embarked"])
```

```python
data
```

```python
x=data.drop("Survived",axis=1)
y=data["Survived"]
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
x_train,x_test,y_train,y_test = tts(x,y,test_size=0.3,random_state=21)
```

```python
from sklearn.linear_model import LogisticRegression
LGR=LogisticRegression
```

```python
model=LGR.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
score=accuracy_score(y_test,y_pred)
score
```

```python
confusion_matrix(y_test,y_pred)
```
